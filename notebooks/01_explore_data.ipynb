hidden blanks check

```python
text_cols = df.select_dtypes(exclude=["number", "bool"]).columns

for c in text_cols:
    s = df[c].dropna().astype(str).str.strip().str.lower()
    hidden = s.isin(["", "na", "n/a", "none", "null", "unknown", "nan"]).sum()
    print(c, hidden)
```
